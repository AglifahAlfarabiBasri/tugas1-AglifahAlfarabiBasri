Tugas 1 Analisis Big Data

Dataset: Wikipedia Indonesia (idwiki) - 20231101
Sumber: HuggingFace (wikimedia/wikipedia)


```python
import polars as pl
import duckdb
import os

# Path ke dataset
DATA_PATH = "../data/raw/idwiki.parquet"

print(f"Ukuran file: {os.path.getsize(DATA_PATH) / (1024*1024):.2f} MB")
```

## 1. Profiling menggunakan Polars


```python
# Menggunakan lazy evaluation dengan scan_parquet
df_lazy = pl.scan_parquet(DATA_PATH)

# Menampilkan skema data
print("Skema Data:")
print(df_lazy.collect_schema())

# Mengambil beberapa baris pertama
print("\nBeberapa baris pertama:")
display(df_lazy.head().collect())

# Menghitung jumlah baris
total_rows = df_lazy.select(pl.len()).collect().item()
print(f"\nTotal baris: {total_rows}")
```

## 2. Profiling menggunakan DuckDB dengan SUMMARIZE


```python
# Menggunakan DuckDB untuk melihat summary statistik dataset
conn = duckdb.connect()

query = f"SUMMARIZE SELECT * FROM '{DATA_PATH}'"
summary_df = conn.execute(query).df()

display(summary_df)
```
